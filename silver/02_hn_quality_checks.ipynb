```python
%sql

CREATE OR REPLACE TABLE
job_market.silver.hn_duplicate_audit
AS
SELECT
    post_id,
    COUNT(*) AS occurrences
FROM job_market.bronze.hacker_news_historical
WHERE post_id IS NOT NULL
GROUP BY post_id
HAVING COUNT(*) > 1;
```

```python
%sql

SELECT
    quarantine_reason,
    COUNT(*) AS total_invalid
FROM job_market.silver.hacker_news_quarantine
GROUP BY quarantine_reason;
```
