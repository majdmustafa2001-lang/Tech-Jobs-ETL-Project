```sql

-- 1. Catalog erstellen
CREATE CATALOG IF NOT EXISTS job_market
MANAGED LOCATION
'abfss://jobs-market@germanelactionstorage.dfs.core.windows.net/uc_managed/';


-- 2. Bronze Schema
CREATE SCHEMA IF NOT EXISTS job_market.bronze
MANAGED LOCATION
'abfss://jobs-market@germanelactionstorage.dfs.core.windows.net/bronze/';


-- 3. Silver Schema
CREATE SCHEMA IF NOT EXISTS job_market.silver
MANAGED LOCATION
'abfss://jobs-market@germanelactionstorage.dfs.core.windows.net/silver/';


-- 4. Gold Schema
CREATE SCHEMA IF NOT EXISTS job_market.gold
MANAGED LOCATION
'abfss://jobs-market@germanelactionstorage.dfs.core.windows.net/gold/';


-- 5. Landing-Schema ohne eigene Managed Location
CREATE SCHEMA IF NOT EXISTS job_market.landing;

-- 6. External Volume für Rohdateien
CREATE EXTERNAL VOLUME IF NOT EXISTS job_market.landing.raw_files
LOCATION 'abfss://jobs-market@germanelactionstorage.dfs.core.windows.net/landing/';

```
